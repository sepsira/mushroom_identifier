```python
# basic print statements to understand shape of df
import pandas as pd

pd.set_option("display.max_rows", None)

df = pd.read_csv("data/training_data.csv")
print(df.shape)
print(df.columns.to_list)
df.head()

df["species"].value_counts()
```

```python
# finding photos of each species! for reference
from PIL import Image
import climage
import shutil


species_names = {
    "Amanita muscaria" : "fly agaric",
    "Trametes versicolor" : "turkey tail",
    "Fomitopsis pinicola" : "red belted conk" ,
    "Hypholoma fasciculare" : "sulphur tuft",
    "Pluteus cervinus" : "deer mushroom",
    "Fomes fomentarius" : "hoof fungus",
    "Auricularia auricula-judae" : "wood ear",
    "Amanita phalloides" : "death cap",
    "Tubaria furfuracea" : "scurfy twiglet",
    "Pleurotus ostreatus" : "oyster mushroom"
}

for species in species_names.keys():
    filtered_df = df[df["species"] == species]
    print(species)

    for i in range(0,5):
        img_id = filtered_df["image_path"].iloc[i].lower()
        img = climage.convert("data/DF20_300/" + img_id, is_unicode=True, is_truecolor=True, is_256color=False, width=shutil.get_terminal_size().columns)

        print(img)

    print("hi")
        
```
